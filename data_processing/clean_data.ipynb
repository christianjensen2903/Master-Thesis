```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
import re
```

```python
# df = pd.read_excel("data/par-to-par-2.xlsx")
df = pd.read_excel("../data/par-to-par-2.xlsx")

for col in ["DATE_FROM", "DATE_TO"]:
    df[col] = pd.to_datetime(df[col], errors="coerce")

df.head()
```

```python
count_before = len(df)

# Remove rows with missing values
df = df.dropna()
print(f"Removed {count_before - len(df)} rows with missing values")

count_before = len(df)

# Remove duplicates
df = df.drop_duplicates()
print(f"Removed {count_before - len(df)} duplicates")
```

```python
# Remove self_case_edges and temporal_violations
count_before = len(df)
df = df[df["CELEX_FROM"] != df["CELEX_TO"]]
df = df[df["DATE_FROM"] >= df["DATE_TO"]]
print(f"Removed {count_before - len(df)} self-citations and temporal violations")
```

```python
# Export to excel
df.to_csv("../data/par-to-par-og.csv", index=False)
```
