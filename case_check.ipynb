```python
import json
import pandas as pd

with open("data/par-to-par-enriched.json", "r", encoding="utf-8") as f:
    data = json.load(f)           # dict: {case_id: {...}, ...}
```

```python
df = pd.read_excel("data/par-to-par-2.xlsx")

for col in ["DATE_FROM", "DATE_TO"]:
    df[col] = pd.to_datetime(df[col], errors="coerce")

df.head()
```

```python
# Get unique CELEX_FROM and CELEX_TO in df as one list
case_ids = list(set(df["CELEX_FROM"].unique().tolist() + df["CELEX_TO"].unique().tolist()))
empty_cases = [case_id for case_id in case_ids if not data.get(case_id, {}).get("meta", {}).get("sections", [])]
non_empty_cases = [case_id for case_id in case_ids if case_id not in set(empty_cases)]
len(empty_cases) / len(case_ids)
```

```python
df[df["CELEX_TO"].isin(non_empty_cases)].sort_values(by="NUMBER_TO", ascending=True)
```

```python
section_counts = {}
for case_id in case_ids:
    if case_id in data:
        case = data[case_id]
        if case.get("meta", {}).get("sections"):
            for section in case["meta"]["sections"]:
                if section not in section_counts:
                    section_counts[section] = 0
                section_counts[section] += 1

# Calculate the fraction of each section type
section_fractions = {section: count / len(case_ids) for section, count in section_counts.items()}
section_fractions
```
